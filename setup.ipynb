```python
!pip install biopython

# 1. Install using the active environment's precise pip path
import sys
!{sys.executable} -m pip install biopython

# 2. Force reload system path
import sys
sys.path.append('/usr/local/lib/python3.10/dist-packages')

# 3. Test the import with proper capitalization (Capital 'B')
import os

from Bio.PDB import PDBParser, PDBList

# 1. Download a sample PDB file (e.g., 4EY7)
pdbl = PDBList()
pdb_file = pdbl.retrieve_pdb_file("4EY7", file_format="pdb", pdir=".")

# Rename file if downloaded in a subfolder format (e.g., pdb1a8m.ent)
for file in os.listdir("."):
    if file.startswith("pdb") and file.endswith(".ent"):
        os.rename(file, "4ey7.pdb")

# 2. Parse the PDB file structure
parser = PDBParser(QUIET=True)
structure = parser.get_structure("protein_sample", "4ey7.pdb")

# We inspect the first model (index 0)
model = structure[0]

# 3. Inspect Chains, Ligands, and Waters
print("--- STRUCTURAL ANALYSIS ---")
for chain in model:
    print(f"\nChain ID: {chain.id}")

    # Track items per chain
    residue_count = 0
    ligands = []
    waters = []

    for residue in chain:
        res_name = residue.get_resname()
        res_id = residue.get_id()

        # Biopython uses the heteroatom flag (res_id[0]) to classify residues:
        # ' ' = Standard amino acid/nucleic acid
        # 'W' = Water molecule
        # 'H_...' = Heteroatom/Ligand

        if res_id[0] == ' ':
            residue_count += 1
        elif res_id[0] == 'W':
            waters.append(f"{res_name}_{res_id[1]}")
        elif res_id[0].startswith('H_'):
            ligands.append(f"{res_name}_{res_id[1]}")

    print(f"  Standard Residues: {residue_count}")
    print(f"  Ligands Found ({len(ligands)}): {', '.join(ligands) if ligands else 'None'}")
    print(f"  Water Molecules Found: {len(waters)}")
```
